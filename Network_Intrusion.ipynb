Comparison

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
from sklearn.preprocessing import label_binarize
import pandas as pd

# --------------------------------------------------
# Predictions from all four models
# --------------------------------------------------

# Logistic Regression
y_pred_lr = logistic_model.predict(X_test_group)

# Decision Tree
y_pred_dt = dt_model.predict(X_test_group)

# Random Forest
y_pred_rf = rf_group.predict(X_test_group)

# XGBoost
y_pred_xgb = xgb_model.predict(X_test_group)


# --------------------------------------------------
# Convert XGBoost numerical predictions back to labels
# --------------------------------------------------

inverse_mapping = {
    0: 'Benign',
    1: 'FTP-BruteForce',
    2: 'SSH-Bruteforce'
}

y_pred_xgb_labels = pd.Series(y_pred_xgb).map(inverse_mapping).values


# --------------------------------------------------
# Function to calculate metrics
# --------------------------------------------------

def evaluate_model(y_true, y_pred, model_name):

    accuracy = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average='macro',
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average='macro',
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average='macro',
        zero_division=0
    )

    return {
        'Model': model_name,
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1
    }


# --------------------------------------------------
# Evaluate all models
# --------------------------------------------------

results = []

results.append(
    evaluate_model(
        y_test_group,
        y_pred_lr,
        'Logistic Regression'
    )
)

results.append(
    evaluate_model(
        y_test_group,
        y_pred_dt,
        'Decision Tree'
    )
)

results.append(
    evaluate_model(
        y_test_group,
        y_pred_rf,
        'Random Forest'
    )
)

results.append(
    evaluate_model(
        y_test_group,
        y_pred_xgb_labels,
        'XGBoost'
    )
)


# --------------------------------------------------
# Create comparison table
# --------------------------------------------------

comparison_df = pd.DataFrame(results)

print(comparison_df.to_string(index=False))
```

```python
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# ABLATION EXPERIMENT
# ============================================================

# Your three dominant features identified through SHAP
dominant_features = [
    "Fwd Seg Size Min",
    "Dst Port",
    "Flow Duration"
]

# ------------------------------------------------------------
# Check that all features actually exist
# ------------------------------------------------------------

for feature in dominant_features:
    if feature not in X_train_group.columns:
        raise ValueError(f"Feature not found: {feature}")

print("All dominant features found.")
print()


# ============================================================
# Define experiments
# ============================================================

experiments = {
    "Full Model (68 features)": [],
    
    "Without Fwd Seg Size Min": [
        "Fwd Seg Size Min"
    ],
    
    "Without Dst Port": [
        "Dst Port"
    ],
    
    "Without Flow Duration": [
        "Flow Duration"
    ],
    
    "Without Fwd Seg Size Min + Dst Port": [
        "Fwd Seg Size Min",
        "Dst Port"
    ],
    
    "Without All 3 Dominant Features": [
        "Fwd Seg Size Min",
        "Dst Port",
        "Flow Duration"
    ]
}


# ============================================================
# Store results
# ============================================================

results = []
models = {}


# ============================================================
# Run each experiment
# ============================================================

for experiment_name, features_to_remove in experiments.items():

    print("\n" + "=" * 80)
    print(experiment_name)
    print("=" * 80)

    # Features to keep
    features_to_keep = [
        col for col in X_train_group.columns
        if col not in features_to_remove
    ]

    X_train_ablation = X_train_group[features_to_keep].copy()
    X_test_ablation = X_test_group[features_to_keep].copy()

    print("Features used:", len(features_to_keep))

    if features_to_remove:
        print("Removed:", features_to_remove)
    else:
        print("Removed: None")

    # --------------------------------------------------------
    # Train a fresh model
    # --------------------------------------------------------

    model_ablation = RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1,
        class_weight="balanced"
    )

    model_ablation.fit(
        X_train_ablation,
        y_train_group
    )

    # --------------------------------------------------------
    # Predictions
    # --------------------------------------------------------

    y_pred = model_ablation.predict(X_test_ablation)

    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_test_group,
        y_pred
    )

    precision = precision_score(
        y_test_group,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_test_group,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_test_group,
        y_pred,
        average="macro",
        zero_division=0
    )

    # --------------------------------------------------------
    # Store results
    # --------------------------------------------------------

    results.append({
        "Experiment": experiment_name,
        "Features": len(features_to_keep),
        "Accuracy": accuracy,
        "Macro Precision": precision,
        "Macro Recall": recall,
        "Macro F1": f1
    })

    # Store model in case we need it later
    models[experiment_name] = model_ablation

    # --------------------------------------------------------
    # Print metrics
    # --------------------------------------------------------

    print(f"\nAccuracy       : {accuracy:.4f}")
    print(f"Macro Precision: {precision:.4f}")
    print(f"Macro Recall   : {recall:.4f}")
    print(f"Macro F1       : {f1:.4f}")

    print("\nClassification Report:")
    print(
        classification_report(
            y_test_group,
            y_pred,
            zero_division=0
        )
    )

    print("Confusion Matrix:")
    print(
        confusion_matrix(
            y_test_group,
            y_pred
        )
    )


# ============================================================
# Final comparison table
# ============================================================

results_df = pd.DataFrame(results)

print("\n")
print("=" * 100)
print("FINAL ABLATION COMPARISON")
print("=" * 100)

print(
    results_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)
```

Temporal Train/test split

```python
import numpy as np
import pandas as pd

# ============================================================
# TEMPORAL HOLDOUT
# Earlier traffic -> TRAIN
# Later traffic  -> TEST
# ============================================================

train_indices = []
test_indices = []

labels = df1["Label"].values

# ------------------------------------------------------------
# 1. SSH-Bruteforce
# ------------------------------------------------------------

ssh_mask = labels == "SSH-Bruteforce"

ssh_times = timestamps[ssh_mask]

ssh_cutoff = ssh_times.quantile(0.70)

ssh_idx = np.where(ssh_mask)[0]

ssh_train_idx = ssh_idx[timestamps.iloc[ssh_idx] < ssh_cutoff]
ssh_test_idx  = ssh_idx[timestamps.iloc[ssh_idx] >= ssh_cutoff]

train_indices.append(ssh_train_idx)
test_indices.append(ssh_test_idx)


# ------------------------------------------------------------
# 2. FTP-BruteForce
# ------------------------------------------------------------

ftp_mask = labels == "FTP-BruteForce"

ftp_times = timestamps[ftp_mask]

ftp_cutoff = ftp_times.quantile(0.70)

ftp_idx = np.where(ftp_mask)[0]

ftp_train_idx = ftp_idx[timestamps.iloc[ftp_idx] < ftp_cutoff]
ftp_test_idx  = ftp_idx[timestamps.iloc[ftp_idx] >= ftp_cutoff]

train_indices.append(ftp_train_idx)
test_indices.append(ftp_test_idx)


# ------------------------------------------------------------
# 3. Benign
# ------------------------------------------------------------

benign_mask = labels == "Benign"

benign_times = timestamps[benign_mask]

benign_cutoff = benign_times.quantile(0.70)

benign_idx = np.where(benign_mask)[0]

benign_train_idx = benign_idx[timestamps.iloc[benign_idx] < benign_cutoff]
benign_test_idx  = benign_idx[timestamps.iloc[benign_idx] >= benign_cutoff]

train_indices.append(benign_train_idx)
test_indices.append(benign_test_idx)


# ------------------------------------------------------------
# Combine indices
# ------------------------------------------------------------

train_indices = np.concatenate(train_indices)
test_indices = np.concatenate(test_indices)


# Sort indices for efficient access
train_indices.sort()
test_indices.sort()


print("=" * 90)
print("TEMPORAL HOLDOUT RESULTS")
print("=" * 90)

print("Training samples:", len(train_indices))
print("Testing samples :", len(test_indices))

print("\nTraining class distribution:")
print(pd.Series(labels[train_indices]).value_counts())

print("\nTesting class distribution:")
print(pd.Series(labels[test_indices]).value_counts())

print("\nCutoff timestamps:")
print("SSH cutoff   :", ssh_cutoff)
print("FTP cutoff   :", ftp_cutoff)
print("Benign cutoff:", benign_cutoff)
```

```python
# ============================================================
# MEMORY-EFFICIENT TEMPORAL ABLATION EXPERIMENT
# ============================================================

import gc
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

ablation_results = []

for experiment_name, selected_features in feature_configs.items():

    print("\n" + "=" * 90)
    print(f"RUNNING: {experiment_name}")
    print(f"Number of features: {len(selected_features)}")
    print("=" * 90)

    # --------------------------------------------------------
    # Extract ONLY the required columns
    # --------------------------------------------------------

    X_train_ablation = X_temp_sample[selected_features]
    X_test_ablation = X_temp_test[selected_features]

    # --------------------------------------------------------
    # Train model
    # --------------------------------------------------------

    model = RandomForestClassifier(
        n_estimators=150,
        random_state=42,
        n_jobs=2,
        class_weight="balanced"
    )

    print("Training...")

    model.fit(
        X_train_ablation,
        y_temp_sample
    )

    # --------------------------------------------------------
    # Prediction
    # --------------------------------------------------------

    print("Predicting...")

    y_pred = model.predict(
        X_test_ablation
    )

    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_temp_test,
        y_pred
    )

    precision = precision_score(
        y_temp_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_temp_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_temp_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    # --------------------------------------------------------
    # Store results
    # --------------------------------------------------------

    ablation_results.append({
        "Experiment": experiment_name,
        "Features": len(selected_features),
        "Accuracy": accuracy,
        "Macro Precision": precision,
        "Macro Recall": recall,
        "Macro F1": f1
    })

    print("\nResults:")
    print(f"Accuracy        : {accuracy:.4f}")
    print(f"Macro Precision : {precision:.4f}")
    print(f"Macro Recall    : {recall:.4f}")
    print(f"Macro F1        : {f1:.4f}")

    # --------------------------------------------------------
    # Free memory BEFORE next experiment
    # --------------------------------------------------------

    del X_train_ablation
    del X_test_ablation
    del y_pred
    del model

    gc.collect()

    print("Memory released.")
```

```python
# ============================================================
# STEP 3 — MEMORY-EFFICIENT SAMPLING
# ============================================================

import numpy as np
import pandas as pd
import gc

RANDOM_STATE = 42
TRAIN_SAMPLE_SIZE = 150_000

# ------------------------------------------------------------
# 1. Get training labels
# ------------------------------------------------------------

y_train_all = df1.iloc[unseen_train_indices]["Label"].to_numpy()

print("=" * 100)
print("AVAILABLE TRAINING DATA")
print("=" * 100)

unique_classes, class_counts = np.unique(
    y_train_all,
    return_counts=True
)

for cls, count in zip(unique_classes, class_counts):
    print(f"{cls:<20}: {count:,}")

# ------------------------------------------------------------
# 2. Stratified sampling
# ------------------------------------------------------------

rng = np.random.default_rng(RANDOM_STATE)

sample_train_indices = []

for cls, count in zip(unique_classes, class_counts):

    # Proportional number of samples for this class
    n_samples = int(
        TRAIN_SAMPLE_SIZE * count / len(y_train_all)
    )

    # Positions inside unseen_train_indices
    class_positions = np.flatnonzero(y_train_all == cls)

    selected_positions = rng.choice(
        class_positions,
        size=n_samples,
        replace=False
    )

    sample_train_indices.extend(
        unseen_train_indices[selected_positions]
    )

# Convert to numpy array
sample_train_indices = np.array(
    sample_train_indices,
    dtype=np.int64
)

# Shuffle training indices
rng.shuffle(sample_train_indices)

# ------------------------------------------------------------
# 3. Extract training features
# ------------------------------------------------------------

X_unseen_train = df1.iloc[
    sample_train_indices
][feature_cols].astype("float32")

y_unseen_train = df1.iloc[
    sample_train_indices
]["Label"].to_numpy()

# ------------------------------------------------------------
# 4. Extract COMPLETE unseen test set
# ------------------------------------------------------------

X_unseen_test = df1.iloc[
    unseen_test_indices
][feature_cols].astype("float32")

y_unseen_test = df1.iloc[
    unseen_test_indices
]["Label"].to_numpy()

# ------------------------------------------------------------
# 5. Display shapes
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("FINAL EXPERIMENT DATA")
print("=" * 100)

print(f"Training features : {X_unseen_train.shape}")
print(f"Training labels   : {y_unseen_train.shape}")

print(f"Testing features  : {X_unseen_test.shape}")
print(f"Testing labels    : {y_unseen_test.shape}")

print("\nTraining distribution:")
print(pd.Series(y_unseen_train).value_counts())

print("\nUnseen test distribution:")
print(pd.Series(y_unseen_test).value_counts())

# ------------------------------------------------------------
# 6. Memory cleanup
# ------------------------------------------------------------

del y_train_all
gc.collect()

print("\nMemory-efficient unseen-session dataset created successfully.")
```

Creating prediction engines

```python
# ============================================================
# PHASE 11 — STEP 11.4
# NIDS PREDICTION ENGINE
# ============================================================

import joblib
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Load production artifacts
# ------------------------------------------------------------

MODEL_PATH = "final_nids_model.pkl"
FEATURE_PATH = "final_nids_features.pkl"

nids_model = joblib.load(MODEL_PATH)
nids_features = joblib.load(FEATURE_PATH)

print("=" * 100)
print("NIDS PREDICTION ENGINE")
print("=" * 100)

print(f"Model loaded      : {MODEL_PATH}")
print(f"Features loaded   : {len(nids_features)}")
print(f"Model type        : {type(nids_model).__name__}")

# ------------------------------------------------------------
# 2. Prediction function
# ------------------------------------------------------------

def predict_network_flow(flow_data):
    """
    Predict the class of a network flow.

    Parameters
    ----------
    flow_data : pandas DataFrame
        DataFrame containing the 68 required network features.

    Returns
    -------
    results : pandas DataFrame
        Prediction, confidence and alert status.
    """

    # --------------------------------------------------------
    # Validate input
    # --------------------------------------------------------

    if not isinstance(flow_data, pd.DataFrame):
        raise TypeError("flow_data must be a pandas DataFrame.")

    missing_features = [
        feature
        for feature in nids_features
        if feature not in flow_data.columns
    ]

    if missing_features:
        raise ValueError(
            f"Missing features: {missing_features}"
        )

    # --------------------------------------------------------
    # Select features in EXACT model order
    # --------------------------------------------------------

    X = flow_data[nids_features].copy()

    # --------------------------------------------------------
    # Predict class
    # --------------------------------------------------------

    predictions = nids_model.predict(X)

    # --------------------------------------------------------
    # Prediction probabilities
    # --------------------------------------------------------

    probabilities = nids_model.predict_proba(X)

    confidence = probabilities.max(axis=1)

    # --------------------------------------------------------
    # Create results
    # --------------------------------------------------------

    results = pd.DataFrame({
        "Prediction": predictions,
        "Confidence": confidence
    })

    # --------------------------------------------------------
    # Generate alert
    # --------------------------------------------------------

    results["Alert"] = np.where(
        results["Prediction"] == "Benign",
        "NORMAL",
        "INTRUSION DETECTED"
    )

    return results


print("\n✓ Prediction engine created successfully.")
```

```python
# ============================================================
# PHASE 11 — STEP 11.5
# NIDS INTRUSION ALERT & LOGGING SYSTEM
# ============================================================

import pandas as pd
import numpy as np
from datetime import datetime
from pathlib import Path

ALERT_LOG_FILE = "nids_alert_log.csv"


# ------------------------------------------------------------
# 1. Severity classification
# ------------------------------------------------------------

def get_severity(prediction, confidence):

    if prediction == "Benign":
        return "NONE"

    if prediction == "SSH-Bruteforce":
        if confidence >= 0.90:
            return "HIGH"
        elif confidence >= 0.70:
            return "MEDIUM"
        else:
            return "LOW"

    if prediction == "FTP-BruteForce":
        if confidence >= 0.90:
            return "HIGH"
        elif confidence >= 0.70:
            return "MEDIUM"
        else:
            return "LOW"

    return "MEDIUM"


# ------------------------------------------------------------
# 2. Generate alerts
# ------------------------------------------------------------

def generate_nids_alerts(prediction_results):

    alerts = []

    for _, row in prediction_results.iterrows():

        prediction = row["Prediction"]
        confidence = float(row["Confidence"])

        # Ignore normal traffic
        if prediction == "Benign":
            continue

        severity = get_severity(
            prediction,
            confidence
        )

        alert = {
            "Timestamp": datetime.now().strftime(
                "%Y-%m-%d %H:%M:%S"
            ),
            "Attack Type": prediction,
            "Confidence": round(confidence, 4),
            "Severity": severity,
            "Status": "INTRUSION DETECTED"
        }

        alerts.append(alert)

        # Display alert
        print("\n" + "!" * 80)
        print("🚨 INTRUSION DETECTED")
        print("!" * 80)
        print(f"Attack Type : {prediction}")
        print(f"Confidence  : {confidence:.2%}")
        print(f"Severity    : {severity}")
        print(f"Timestamp   : {alert['Timestamp']}")

    # --------------------------------------------------------
    # No attacks
    # --------------------------------------------------------

    if not alerts:
        print("\n✓ No intrusion detected.")

        return pd.DataFrame(
            columns=[
                "Timestamp",
                "Attack Type",
                "Confidence",
                "Severity",
                "Status"
            ]
        )

    return pd.DataFrame(alerts)


# ------------------------------------------------------------
# 3. Save alerts
# ------------------------------------------------------------

def save_alerts(alerts):

    if alerts.empty:
        print("\nNo alerts to save.")
        return

    file_exists = Path(ALERT_LOG_FILE).exists()

    alerts.to_csv(
        ALERT_LOG_FILE,
        mode="a",
        header=not file_exists,
        index=False
    )

    print(
        f"\n✓ {len(alerts)} alert(s) saved to "
        f"{ALERT_LOG_FILE}"
    )


print("=" * 100)
print("NIDS ALERT & LOGGING SYSTEM")
print("=" * 100)

print("✓ Alert generation function created.")
print("✓ Severity classification created.")
print("✓ Alert logging system created.")
```

```python
# ============================================================
# PHASE 11 — STEP 11.6
# STREAMING NIDS SIMULATOR
# ============================================================

import time
import pandas as pd
import numpy as np

print("=" * 100)
print("STREAMING NIDS SIMULATOR")
print("=" * 100)


def process_network_flow(flow, flow_id):
    """
    Process one network-flow record through the NIDS.
    """

    # Convert single row into DataFrame
    flow_df = pd.DataFrame(
        [flow]
    )

    # Run prediction
    result = predict_network_flow(
        flow_df
    )

    prediction = result.iloc[0]["Prediction"]
    confidence = float(
        result.iloc[0]["Confidence"]
    )

    # --------------------------------------------------------
    # Display result
    # --------------------------------------------------------

    print("\n" + "-" * 80)
    print(f"FLOW #{flow_id}")
    print("-" * 80)

    print(f"Prediction : {prediction}")
    print(f"Confidence : {confidence:.2%}")

    # --------------------------------------------------------
    # Generate alert if attack
    # --------------------------------------------------------

    if prediction != "Benign":

        alerts = generate_nids_alerts(
            result
        )

        save_alerts(
            alerts
        )

    else:

        print("Status     : ✓ NORMAL TRAFFIC")


print("\n✓ Streaming NIDS function created.")
```

```python
# ============================================================
# PHASE 11 — STEP 11.7
# END-TO-END NIDS EVALUATION
# ============================================================

import time
import pandas as pd
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

print("=" * 100)
print("END-TO-END NIDS EVALUATION")
print("=" * 100)

# ------------------------------------------------------------
# 1. Start timing
# ------------------------------------------------------------

start_time = time.perf_counter()

# ------------------------------------------------------------
# 2. Run the complete prediction pipeline
# ------------------------------------------------------------

y_e2e_pred = predict_network_flow(
    X_test_group
)["Prediction"].to_numpy()

end_time = time.perf_counter()

# ------------------------------------------------------------
# 3. Calculate detection metrics
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_test_group,
    y_e2e_pred
)

precision = precision_score(
    y_test_group,
    y_e2e_pred,
    average="macro",
    zero_division=0
)

recall = recall_score(
    y_test_group,
    y_e2e_pred,
    average="macro",
    zero_division=0
)

f1 = f1_score(
    y_test_group,
    y_e2e_pred,
    average="macro",
    zero_division=0
)

# ------------------------------------------------------------
# 4. Calculate processing performance
# ------------------------------------------------------------

total_samples = len(X_test_group)
total_time = end_time - start_time

flows_per_second = total_samples / total_time
milliseconds_per_flow = (
    total_time / total_samples
) * 1000

# ------------------------------------------------------------
# 5. Count errors
# ------------------------------------------------------------

errors = np.sum(
    y_test_group.to_numpy() != y_e2e_pred
)

correct = total_samples - errors

# ------------------------------------------------------------
# 6. Display overall results
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("OVERALL NIDS PERFORMANCE")
print("=" * 100)

print(f"Total flows tested       : {total_samples:,}")
print(f"Correct detections       : {correct:,}")
print(f"Incorrect detections     : {errors:,}")

print(f"\nAccuracy                 : {accuracy:.4f}")
print(f"Macro Precision          : {precision:.4f}")
print(f"Macro Recall             : {recall:.4f}")
print(f"Macro F1                 : {f1:.4f}")

print("\n" + "=" * 100)
print("PROCESSING PERFORMANCE")
print("=" * 100)

print(f"Total processing time    : {total_time:.4f} seconds")
print(f"Flows processed/second   : {flows_per_second:,.2f}")
print(f"Time per flow            : {milliseconds_per_flow:.4f} ms")

# ------------------------------------------------------------
# 7. Per-class detection performance
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("PER-CLASS DETECTION PERFORMANCE")
print("=" * 100)

report = classification_report(
    y_test_group,
    y_e2e_pred,
    digits=4,
    zero_division=0
)

print(report)

# ------------------------------------------------------------
# 8. Confusion matrix
# ------------------------------------------------------------

labels = [
    "Benign",
    "FTP-BruteForce",
    "SSH-Bruteforce"
]

cm = confusion_matrix(
    y_test_group,
    y_e2e_pred,
    labels=labels
)

cm_df = pd.DataFrame(
    cm,
    index=[f"Actual: {x}" for x in labels],
    columns=[f"Predicted: {x}" for x in labels]
)

print("=" * 100)
print("CONFUSION MATRIX")
print("=" * 100)

print(cm_df)

# ------------------------------------------------------------
# 9. Security-specific metrics
# ------------------------------------------------------------

# False positives = benign traffic classified as attack
benign_index = labels.index("Benign")

false_positives = (
    cm[benign_index].sum()
    - cm[benign_index, benign_index]
)

# False negatives = attacks classified as benign
attack_indices = [
    labels.index("FTP-BruteForce"),
    labels.index("SSH-Bruteforce")
]

false_negatives = cm[
    attack_indices,
    benign_index
].sum()

actual_attacks = cm[
    attack_indices,
    :
].sum()

actual_benign = cm[
    benign_index,
    :
].sum()

attack_detection_rate = (
    (actual_attacks - false_negatives)
    / actual_attacks
)

false_positive_rate = (
    false_positives / actual_benign
)

print("\n" + "=" * 100)
print("SECURITY-SPECIFIC METRICS")
print("=" * 100)

print(f"Actual attack flows      : {actual_attacks:,}")
print(f"Actual benign flows      : {actual_benign:,}")

print(f"False positives          : {false_positives:,}")
print(f"False negatives          : {false_negatives:,}")

print(
    f"Attack Detection Rate    : "
    f"{attack_detection_rate:.4f}"
)

print(
    f"False Positive Rate      : "
    f"{false_positive_rate:.6f}"
)

print("\n" + "=" * 100)
print("STEP 11.7 COMPLETE")
print("=" * 100)
```

Recreating Temporal Holdouts

```python
import numpy as np
import pandas as pd

# ============================================================
# RECREATE CLASS-AWARE TEMPORAL HOLDOUT INDICES
# ============================================================

timestamps = pd.to_datetime(
    df1["Timestamp"],
    format="%d/%m/%Y %H:%M:%S",
    errors="coerce"
)

labels = df1["Label"]

classes = [
    "Benign",
    "FTP-BruteForce",
    "SSH-Bruteforce"
]

temporal_train_indices = []
temporal_test_indices = []

for cls in classes:

    class_indices = np.where(labels.values == cls)[0]

    class_times = timestamps.iloc[class_indices]

    cutoff = class_times.quantile(0.70)

    train_idx = class_indices[class_times <= cutoff]
    test_idx = class_indices[class_times > cutoff]

    temporal_train_indices.extend(train_idx)
    temporal_test_indices.extend(test_idx)

temporal_train_indices = np.array(temporal_train_indices)
temporal_test_indices = np.array(temporal_test_indices)

print("Temporal training rows:", len(temporal_train_indices))
print("Temporal testing rows :", len(temporal_test_indices))
```

Create the same 100k/50k samples

```python
# ============================================================
# STRATIFIED TEMPORAL SAMPLING
# ============================================================

rng = np.random.default_rng(42)

train_sample_size = 100000
test_sample_size = 50000

# ------------------------------------------------------------
# TRAIN SAMPLE
# ------------------------------------------------------------

train_sample_indices = []

for cls in classes:

    cls_indices = temporal_train_indices[
        labels.iloc[temporal_train_indices].values == cls
    ]

    proportion = len(cls_indices) / len(temporal_train_indices)

    n_samples = int(round(train_sample_size * proportion))

    selected = rng.choice(
        cls_indices,
        size=n_samples,
        replace=False
    )

    train_sample_indices.extend(selected)

train_sample_indices = np.array(train_sample_indices)


# ------------------------------------------------------------
# TEST SAMPLE
# ------------------------------------------------------------

test_sample_indices = []

for cls in classes:

    cls_indices = temporal_test_indices[
        labels.iloc[temporal_test_indices].values == cls
    ]

    proportion = len(cls_indices) / len(temporal_test_indices)

    n_samples = int(round(test_sample_size * proportion))

    selected = rng.choice(
        cls_indices,
        size=n_samples,
        replace=False
    )

    test_sample_indices.extend(selected)

test_sample_indices = np.array(test_sample_indices)


print("Temporal train sample:", len(train_sample_indices))
print("Temporal test sample :", len(test_sample_indices))
```

```python
# ============================================================
# HANDLE SHAP OUTPUT FORMAT
# ============================================================

print("SHAP shape:", np.array(shap_values_40).shape)

# For multiclass Random Forest, SHAP can have different
# output formats depending on SHAP version.

if isinstance(shap_values_40, list):

    # Old SHAP format:
    # [class0_array, class1_array, class2_array]

    mean_abs_shap_40 = np.mean(
        [
            np.abs(values).mean(axis=0)
            for values in shap_values_40
        ],
        axis=0
    )

else:

    shap_array_40 = np.array(shap_values_40)

    # Expected modern format:
    # samples × features × classes

    if shap_array_40.ndim == 3:

        mean_abs_shap_40 = np.abs(
            shap_array_40
        ).mean(axis=(0, 2))

    elif shap_array_40.ndim == 2:

        mean_abs_shap_40 = np.abs(
            shap_array_40
        ).mean(axis=0)

    else:
        raise ValueError(
            f"Unexpected SHAP shape: {shap_array_40.shape}"
        )

# ------------------------------------------------------------
# 4. Create feature importance table
# ------------------------------------------------------------

shap_importance_40 = pd.DataFrame({
    "Feature": X_shap_40.columns,
    "Mean |SHAP|": mean_abs_shap_40
})

shap_importance_40 = (
    shap_importance_40
    .sort_values("Mean |SHAP|", ascending=False)
    .reset_index(drop=True)
)

print("\n" + "="*60)
print("TOP 20 SHAP FEATURES — 40 FEATURE MODEL")
print("="*60)

display(
    shap_importance_40.head(20).round(6)
)
```

```python
def predict_network_flow_40(flow_data):

    # --------------------------------------------------------
    # Validate input
    # --------------------------------------------------------

    if not isinstance(flow_data, pd.DataFrame):
        raise TypeError(
            "flow_data must be a pandas DataFrame."
        )

    # --------------------------------------------------------
    # Check required features
    # --------------------------------------------------------

    missing_features = [
        feature
        for feature in final_features_40
        if feature not in flow_data.columns
    ]

    if missing_features:
        raise ValueError(
            f"Missing required features: {missing_features}"
        )

    # --------------------------------------------------------
    # Select features in EXACT model order
    # --------------------------------------------------------

    X = flow_data[final_features_40].copy()

    # --------------------------------------------------------
    # Prediction
    # --------------------------------------------------------

    predictions = final_model_40.predict(X)

    probabilities = final_model_40.predict_proba(X)

    confidence = probabilities.max(axis=1)

    # --------------------------------------------------------
    # Result dataframe
    # --------------------------------------------------------

    results = pd.DataFrame({
        "Prediction": predictions,
        "Confidence": confidence
    })

    # --------------------------------------------------------
    # Alert classification
    # --------------------------------------------------------

    results["Alert"] = np.where(
        results["Prediction"] == "Benign",
        "NORMAL",
        "INTRUSION DETECTED"
    )

    return results
```
